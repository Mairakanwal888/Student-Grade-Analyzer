```python
import numpy as np

# Student names
students = np.array(["Ali", "Sara", "Ahmed", "Ayesha", "Hamza"])

# Marks: rows = students, columns = subjects
# Subjects: Math, English, Science
marks = np.array([
    [85, 78, 92],
    [72, 88, 81],
    [90, 95, 89],
    [65, 70, 68],
    [78, 82, 75]
])

# Calculate total marks for each student
total = np.sum(marks, axis=1)

# Calculate average marks for each student
average = np.mean(marks, axis=1)

# Find highest and lowest marks
highest = np.max(marks, axis=1)
lowest = np.min(marks, axis=1)

# Display results
print("===== Student Grade Analyzer =====\n")

for i in range(len(students)):
    print(f"Student: {students[i]}")
    print(f"Marks: {marks[i]}")
    print(f"Total: {total[i]}")
    print(f"Average: {average[i]:.2f}")
    print(f"Highest Mark: {highest[i]}")
    print(f"Lowest Mark: {lowest[i]}")

    if average[i] >= 50:
        print("Status: PASS")
    else:
        print("Status: FAIL")

    print("-------------------------")

# Class statistics
print("\n===== Class Statistics =====")

print("Class Average:", np.mean(average))
print("Highest Mark in Class:", np.max(marks))
print("Lowest Mark in Class:", np.min(marks))

# Find the student with the highest average
top_student_index = np.argmax(average)

print(
    "Top Student:",
    students[top_student_index],
    "with average",
    average[top_student_index]
)
```
